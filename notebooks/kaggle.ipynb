# nanoscope on Kaggle

Runs any study file on both T4s, one run per GPU at a time. Set `STUDY` and `RUNS_REPO` in the first cell.

Before running:
- **Accelerator:** set it to *GPU T4 x2*.
- **Internet:** turn it on.
- **Secret:** add `HF_TOKEN` (Add-ons → Secrets) with a Hugging Face token that has write access.

Every run is mirrored to a private Hub repo at its checkpoints. If the 12-hour session ends, start a new one and run the same cells: finished runs are skipped and unfinished ones resume from their last checkpoint.

```python
STUDY = "studies/m1_ablation.py"      # any study file in the repo
RUNS_REPO = "your-hf-username/nanoscope-runs"  # private Hub repo that keeps runs between sessions
BRANCH = "main"

!git clone -q --depth 1 -b {BRANCH} https://github.com/almajd3713/nanoscope /kaggle/working/nanoscope
!pip install -q -e "/kaggle/working/nanoscope[wandb]"
```

```python
import os
from kaggle_secrets import UserSecretsClient

os.environ["HF_TOKEN"] = UserSecretsClient().get_secret("HF_TOKEN")
os.chdir("/kaggle/working/nanoscope")
```

```python
!nanoscope study {STUDY} --devices cuda:0,cuda:1 --push-to-hub {RUNS_REPO}
```

`nanoscope study` prints its progress every 30 seconds and the report when it finishes. From a second notebook or a terminal, `nanoscope status runs` shows every run's step and latest validation result.
